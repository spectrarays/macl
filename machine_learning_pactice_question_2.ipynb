```python
from sklearn.datasets import load_iris
import pandas as pd


iris = load_iris()
```

```python
X = pd.DataFrame(iris["data"])
Y = pd.DataFrame(iris["target"])
```

0 -> Sepal Length,

1 -> Sepal Width,

2 -> Petal Length,

3 -> Petal Width.

## **Descriptive analysis**

```python
X.describe()
```

```python
X.mean()
```

```python
X.median()
```

```python
X.mode().T  #feature 2 have two modes
```

```python
X.std()
```

```python
X.sem()
```

```python
print(f"95% Confidence Interval using z score is:\n{X.mean()+(1.96*X.sem())}")
print(f"95% Confidence Interval using z score is:\n{X.mean()-(1.96*X.sem())}")
```

```python
corr_matrix = X.corr(method="pearson")

import seaborn as sns

sns.heatmap(corr_matrix, annot=True)
```

## **Covariance between petal length and sepal length**

```python
print(f"Covariance between length of petal and sepal is: {X[0].cov(X[2])}")
```

## **Contingency Table**

```python
pd.crosstab(Y[0], Y[0])
```
